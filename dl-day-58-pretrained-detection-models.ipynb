# Why Start With Pretrained Models?

Before training detectors yourself, you must understand:

- What outputs look like
- How predictions are represented
- What good predictions look like
- What bad predictions look like

Think of it like this:

**Classification:**
Image → Class

**Detection:**
Image → Multiple Objects + Locations

Today's goal is observation, not training.

# What Is YOLO?

**YOLO = You Only Look Once**

YOLO is a family of real-time object detection models designed to detect objects in a single forward pass through a neural network.

Popular versions include:

- YOLOv5 (2020)
- YOLOv8 (2023)
- YOLOv9 (2024)
- YOLOv10 (2024)
- YOLOv11 (2024)

## Core Idea

Traditional detectors often use:

```markdown
Image
↓
Find candidate regions
↓
Classify each region
↓
Predict boxes
```
YOLO instead does:

```markdown
Image
↓
Single CNN Forward Pass
↓
Predict all objects directly
```

This is why YOLO is extremely fast.

## Why "You Only Look Once"?

The model processes the image:

> one time

and simultaneously predicts:

- object classes
- bounding boxes
- confidence scores

No separate region proposal stage is needed.

## What YOLO Predicts

For each detected object, YOLO predicts:

1. Class
2. Confidence Score
3. Bounding Box

**Example Detection**

```markdown
Class      : Person
Confidence : 0.96
```

Bounding Box:

```
x1 = 120
y1 = 50
x2 = 310
y2 = 450
```

Meaning:

A person is detected inside this rectangle with 96% confidence.


**Detector Output Structure**

A typical detection output:


$$[class, confidence, x1, y1, x2, y2]$$

**Example**

$$
[
  person,
  0.95,
  100,
  50,
  300,
  400
]$$


<u>Interpretation</u>

```markdown
Class      = Person
Confidence = 95%
```
Bounding Box:

```markdown
Top-left     = (100, 50)
Bottom-right = (300, 400)
```

Meaning:

A person exists in this region with high confidence.

## Complete YOLO Pipeline

```text
Input Image
↓
CNN Backbone
↓
Feature Maps
↓
Detection Head
↓
Class Scores + Bounding Boxes
↓
Non-Maximum Suppression (NMS)
↓
Final Detections
```


## Why YOLO Became Popular

### Advantages

✔ Real-time inference

✔ End-to-end detection

✔ Excellent speed

✔ Easy deployment

✔ Strong speed-to-accuracy trade-off


### Limitations

❌ Small object detection can be challenging

❌ May be slightly less accurate than large two-stage detectors in some cases

❌ Requires high-quality annotations
